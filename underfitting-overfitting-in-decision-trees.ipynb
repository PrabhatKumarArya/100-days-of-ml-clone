```python
from sklearn.tree import DecisionTreeClassifier
def analyser(max_depth):
    df = pd.read_csv(r"\Users\Anjli Kumari\campusx-lectures\datasets\Social_Network_Ads.csv")
    X = df.iloc[:,2:4].values
    Y = df.iloc[:,-1].values
    clf = DecisionTreeClassifier(max_depth=max_depth)
    clf.fit(X,Y)
    a = np.arange(start=X[:,0].min()-1,stop=X[:,0].max()+1,step=0.1)
    b = np.arange(start=X[:,1].min()-1,stop=X[:,1].max()+1,step=100)
    XX,YY = np.meshgrid(a,b)
    inp_arr = np.array([XX.ravel(),YY.ravel()]).T
    labels = clf.predict(inp_arr)
    plt.contourf(XX,YY,labels.reshape(XX.shape),alpha=0.5)
    plt.scatter(X[:,0],X[:,1],c=Y)
```

```python
#overfitting
analyser(max_depth=None)
```

```python
#underfitting
analyser(max_depth=5)
```
