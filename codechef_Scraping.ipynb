```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re

profile_names = []
user_names = []
leagues = []
countries = []
ratings  = []
contest_ratings = []
contests = []
solved = []

url = "https://www.codechef.com/users/tourist"

headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}

users = requests.get(url,headers=headers).text
soup = BeautifulSoup(users,'lxml')
soup.prettify()

for i in soup.find_all('div', class_='user-profile-container'):

    # Profile Name
    tag = i.find('h1', class_='h2-style')
    profile_names.append(tag.get_text(strip=True) if tag else None)

    # Username
    tag = i.find('span', class_='m-username--link')
    user_names.append(tag.get_text(strip=True) if tag else None)

    # Rating
    tag = i.find('span', class_='rating')
    ratings.append(tag.get_text(strip=True).split()[0] if tag else None)

    # Country
    tag = i.find('span', class_='user-country-name')
    countries.append(tag.get_text(strip=True) if tag else None)

    # Contest Rating
    tag = i.find('a', class_='rating')
    contest_ratings.append(tag.get_text(strip=True).split()[0] if tag else None)

    # League
    tag = i.find('span', class_='tooltip')
    leagues.append(tag.get_text(strip=True).split()[0] if tag else None)

    # Problems Solved Section
    info = i.find('section', class_='rating-data-section problems-solved')

    if info:
        props = [h3.get_text(strip=True) for h3 in info.find_all('h3')]

        contests.append(int(re.search(r'\d+', props[-2]).group()) if len(props) >= 2 else None)
        solved.append(int(re.search(r'\d+', props[-1]).group()) if len(props) >= 1 else None)
    else:
        contests.append(None)
        solved.append(None)

df = pd.DataFrame({
    "Profile-Name": profile_names,
    "Username": user_names,
    "League": leagues,
    "Country": countries,
    "User-Rating": ratings,
    "Current-Contest-Rating": contest_ratings,
    "No-Of-Contests-Participated": contests,
    "Total-Problems-Solved": solved
})

print(df)
```
