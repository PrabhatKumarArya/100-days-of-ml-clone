### Iterative Imputer Using Scikit-learn

```python
df = pd.read_csv(r"\Users\Anjli Kumari\datasets\50_Startups.csv")
df.head(1)
```

```python
df['ProfitClass'] = (df['Profit'] >= df['Profit'].median()).astype(int)
df.head()
```

```python
X = df.iloc[:,0:3]
Y = df.iloc[:,-1]
Y
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
X_train,X_test,Y_train,Y_test = train(X,Y,0.2,42)
imp = IterativeImputer(estimator=LinearRegression(),max_iter=10,random_state=42)
X_train_trf = imp.fit_transform(X_train)
X_test_trf = imp.transform(X_test)
lr = LinearRegression()
lr.fit(X_train_trf,Y_train)
Y_pred = lr.predict(X_test_trf)
print("R2 Score:",r2_score(Y_test,Y_pred))
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
X_train,X_test,Y_train,Y_test = train(X,Y,0.2,42)
imp = IterativeImputer(estimator=LinearRegression(),max_iter=13,random_state=42)
X_train_trf = imp.fit_transform(X_train)
X_test_trf = imp.transform(X_test)
lor = LogisticRegression(max_iter=1000)
lor.fit(X_train_trf,Y_train)
Y_pred = lor.predict(X_test_trf)
print("Accuracy:",accuracy_score(Y_test,Y_pred))
```

```python
for seed in [1, 10, 20, 30, 40, 42, 50, 100]:
    X_train, X_test, Y_train, Y_test = train_test_split(
        X, Y,
        test_size=0.2,
        random_state=seed,
        stratify=Y
    )

    imp = IterativeImputer(
        estimator=LinearRegression(),
        max_iter=10,
        random_state=42
    )

    X_train_trf = imp.fit_transform(X_train)
    X_test_trf = imp.transform(X_test)

    lor = LogisticRegression(max_iter=1000)
    lor.fit(X_train_trf, Y_train)

    Y_pred = lor.predict(X_test_trf)

    print(seed, accuracy_score(Y_test, Y_pred))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(Y_test, Y_pred))
print(classification_report(Y_test, Y_pred))
```
