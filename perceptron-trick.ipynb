```python
from sklearn.datasets import make_classification
X,Y = make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,class_sep=20)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c = Y,cmap='winter',s=100)
```

```python
def step(z):
    return 1 if z > 0 else 0
```

```python
def perceptron(X,Y):
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    learning_rate = 0.1

    for i in range(1000):
        j = np.random.randint(0,100)
        Y_hat = step(np.dot(X[j],weights))
        weights = weights + learning_rate * ((Y[j]-Y_hat) * X[j])

    return weights[0],weights[1:]
```

```python
intercept_,coef_ = perceptron(X,Y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
X_inp = np.linspace(-3,3,100)
Y_inp = m * X_inp + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(X_inp,Y_inp,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=Y,cmap='winter',s=100)
plt.ylim(-3,2)
```

```python
def perceptron(X,Y):
    m,b = [],[]
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    learning_rate = 0.1

    for i in range(200):
        j = np.random.randint(0,100)
        Y_hat = step(np.dot(X[j],weights))
        weights = weights + learning_rate * (Y[j]-Y_hat) * X[j]
        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))

    return m,b
```

```python
m,b = perceptron(X,Y)
```

### Perceptron Using Sigmoid Function

```python
from sklearn.datasets import make_classification
X,Y = make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,class_sep=30)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=Y,cmap='winter',s=100)
plt.show()
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def perceptron(X,Y):
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    learning_rate = 0.1

    for i in range(1000):
        j = np.random.randint(0,100)
        Y_hat = sigmoid(np.dot(X[j],weights))
        weights = weights + learning_rate * (Y[j] - Y_hat) * X[j]

    return weights[0],weights[1:]
```

```python
intercept_,coef_ = perceptron(X,Y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
X_inp2 = np.linspace(-3,3,100)
Y_inp2 = m * X_inp + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(X_inp,Y_inp,color='red',linewidth=3,label='Perceptron-Step')
plt.plot(X_inp1,Y_inp1,color='black',linewidth=3,label='Scikit-Learn')
plt.plot(X_inp2,Y_inp2,color='brown',linewidth=3,label='Perceptron-Sigmoid')
plt.scatter(X[:,0],X[:,1],c=Y,cmap='winter',s=100)
plt.ylim(-3,2)
plt.legend()
plt.show()
```
