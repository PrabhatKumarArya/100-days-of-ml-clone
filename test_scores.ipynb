```python
import numpy as np
import matplotlib.pyplot as plt
import math

def gradient_function(x, y):
    m_curr = b_curr = 0
    n = len(x)
    iterations = 10
    learning_rate = 0.003

    for i in range(iterations):
        y_predicted = m_curr * x + b_curr
        cost = (1/n) * sum(val ** 2 for val in (y - y_predicted))

        md = -(2/n) * sum(x * (y - y_predicted))
        bd = -(2/n) * sum(y - y_predicted)

        m_prev = m_curr
        b_prev = b_curr

        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd

        if i > 0 and math.isclose(m_curr, m_prev, rel_tol=1e-6) and math.isclose(b_curr, b_prev, rel_tol=1e-6):
            print(f"Converged at iteration {i+1}")
            break

        if i % 100 == 0
            print("Iteration {}, m: {}, b: {}, cost: {}".format(i+1,m_curr,b_curr,cost))

        return m_curr,b_curr

x = np.array([92, 56, 88, 70, 80, 49, 65, 35, 66, 67])
y = np.array([98, 68, 81, 80, 83, 52, 66, 30, 68, 73])

m,b = gradient_function(x, y)
print(f"Final values: m = {m}, b = {b}")

plt.scatter(x, y)
plt.plot(x, m*x + b, color='red')
plt.show()
```
