```python
from collections import deque
```

```python
q = None
vis = None
path = None
prev = None
def bfs(graph,start,goal):
    # neighbour list of each node
    q = deque([[start]])
    # start with root node
    vis = set([start])
    while q:
        # checking for each path
        path = q.popleft()
        # tracking dest of each path
        node = path[-1]
        # goal-test
        if node == goal:
            return path

        # explore neighbours if goal-test fails
        for neighbour in graph.get(node,[]):
            if neighbour not in vis:
                vis.add(neighbour)
                q.append(path+[neighbour])
    return None  
```

```python
n = int(input("Enter Number of Edges:"))
graph = {}

for _ in range(n):
    u,v = input("Enter Edge u v : ").split()
    if u not in graph: graph[u] = []
    if v not in graph : graph[v] = []
    graph[u].append(v)
    graph[v].append(u)

start = input("Enter Start Node:")
goal = input("Enter Goal Node:")

path = bfs(graph,start,goal)
print("Path:" , path if path else "No Path Found")
```

```python
def dfs(graph,node,goal,vis=None,path=None):
    if vis is None: vis = set()
    if path is None : path = []
    # start with root node
    vis.add(node)
    path.append(node)
    # goal-test satisfies
    if node == goal:
        return path
    # explore neighbours if goal-test fails
    for neighbour in graph.get(node,[]):
        if neighbour not in vis:
            # apply goal-test for each neighbour
            res = dfs(graph,node=neighbour,goal=goal,vis=vis,path=path.copy())
            # return path which succeeds first goal-test
            if res:
                return res
    return None
```

```python
n = int(input("Enter Number of Edges:"))
graph = {}

for _ in range(n):
    u,v = input("Enter edges u,v: ").split()
    if u not in graph:graph[u] = []
    if v not in graph: graph[v] = []
    graph[u].append(v)
    graph[v].append(u)

start = input("Enter Start Node:")
goal = input("Enter Goal Node:")

path = dfs(graph,start,goal)
print("Path:",path if path else "Path Not Found")
```
