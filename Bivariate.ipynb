```python
#numerical-numerical(scatterplot)
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='smoker',size='size')
plt.show()
```

```python
#numerical-categorical(barplot)
sns.barplot(data=titanic,x='Pclass',y='Fare',hue='Sex',palette=['red','green'])
plt.show()
```

```python
#numerical-categorical(boxplot)
sns.boxplot(data=titanic,x='Sex',y='Age',hue='Survived',palette=['red','blue'])
plt.show()
```

```python
?sns.displot
```

```python
#numerical-categorical(displot)
sns.displot(data=titanic,x='Sex',y='Age',hue='SibSp',kind='hist',palette=['red','orange','green','yellow','blue','violet','indigo'])
plt.show()
```

```python
?sns.heatmap
```

```python
?pd.crosstab
```

```python
?sns.heatmap
```

```python
sns.heatmap(pd.crosstab(index=[titanic['Pclass']],columns=[titanic['Survived'],titanic['Sex']],normalize='index'),annot=True,fmt='.2f',cmap='viridis')
plt.show()
```

```python
#c-chevauk,q-queenstown,s-southhampean
table = pd.crosstab(index=[titanic['Pclass'],titanic['Survived']],columns=[titanic['Sex'],titanic['Embarked']],normalize='index') * 100
labels = table.round(1).astype(str) + "%"
sns.heatmap(table,annot=labels,fmt="",cmap="viridis")
plt.show()
```

```python
?sns.clustermap
```

```python
#categoriacal-categorical
sns.clustermap(pd.crosstab(index=titanic['Parch'],columns=titanic['Survived']))
plt.show()
```
