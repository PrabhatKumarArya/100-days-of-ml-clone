#### Comparing Imputers

| Method | Advantages | Disadvantages |
|--------|------------|---------------|
| **Mean Imputation** | • Simple and fast<br>• Easy to implement<br>• Suitable for normally distributed numerical data | • Sensitive to outliers<br>• Reduces data variance<br>• Can distort the original distribution |
| **Median Imputation** | • Robust to outliers<br>• Good for skewed numerical data<br>• Preserves the central tendency better than mean | • Ignores relationships between features<br>• May not be ideal for normally distributed data |
| **Mode (Most Frequent) Imputation** | • Suitable for categorical data<br>• Simple and fast<br>• Preserves valid category values | • Can over-represent the most frequent category<br>• May introduce bias if one category dominates |
| **KNN Imputation** | • Uses information from similar samples<br>• Preserves relationships between features<br>• Often provides more accurate imputations | • Computationally expensive<br>• Slower for large datasets<br>• Sensitive to feature scaling<br>• Works only with numerical features |

| Method | Advantages | Disadvantages |
|---|---|---|
| Label Encoding | Simple and fast; Low memory; Good for target labels | Creates artificial order; Not good for nominal data |
| One-Hot Encoding | No artificial order; Good for nominal data; Easy to understand | Increases columns; High memory for many categories |
| Ordinal Encoding | Preserves natural order; Simple; Low memory | Requires meaningful order; Wrong order can mislead model |
| Target Encoding | Reduces dimensions; Good for high-cardinality data | Risk of data leakage; Can overfit |
| Frequency Encoding | Simple and fast; Handles many categories; Low memory | Loses category identity; Different categories may have same frequency |
| Binary Encoding | Fewer columns than OHE; Good for high-cardinality data | Less interpretable; More complex |

##### Outlier Removal vs Capping

**Outlier Removal**

Advantages:
- Simple and easy to implement
- Removes extreme values
- Does not change remaining values
- Can improve model performance

Disadvantages:
- Reduces dataset size
- May remove useful information
- Can introduce bias
- Not suitable for genuine outliers


**Outlier Capping**

Advantages:
- Keeps all observations
- Preserves dataset size
- Reduces effect of extreme values
- Less information is lost

Disadvantages:
- Changes original values
- Reduces data variability
- May hide genuine extreme values
- Choice of cap can affect results

```python
fig, axes = plt.subplots(3, 1, figsize=(42,15))
sns.boxplot(x=original, color='Green', ax=axes[0])
axes[0].set_title(f"Original Data Of {col} Column")
sns.boxplot(x=standard, color='Blue', ax=axes[1])
axes[1].set_title(f"StandardScaler Data Of {col} Column")
sns.boxplot(x=minmax, color='Purple', ax=axes[2])
axes[2].set_title(f"MinMaxScaler Data Of {col} Column")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(40,15))
sns.histplot(original, kde=True,color='Orange', ax=axes[0])
axes[0].set_title(f"Original Data Of {col} Column")
sns.histplot(standard, kde=True,color='Magenta', ax=axes[1])
axes[1].set_title(f"StandardScaler Data Of {col} Column")
sns.histplot(minmax, kde=True,color='ForestGreen', ax=axes[2])
axes[2].set_title(f"MinMaxScaler Data Of {col} Column")
plt.tight_layout()
plt.show()
```
